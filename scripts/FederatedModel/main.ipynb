```python
def create_image_dashboard(title, image_paths, columns=3):
    """
    Crea un dashboard HTML con imágenes PNG
    
    Args:
        title (str): Título del dashboard
        image_paths (list): Lista de rutas de las imágenes PNG
        columns (int): Número de columnas para mostrar las imágenes
    """
    html_content = f"""
    <!DOCTYPE html>
    <html>
    <head>
        <title>{title}</title>
        <style>
            .dashboard {{
                max-width: 1200px;
                margin: 0 auto;
                padding: 20px;
            }}
            .image-grid {{
                display: grid;
                grid-template-columns: repeat({columns}, 1fr);
                gap: 20px;
            }}
            .image-container {{
                text-align: center;
            }}
            img {{
                max-width: 100%;
                height: auto;
                border: 1px solid #ddd;
                border-radius: 8px;
            }}
            h1 {{
                text-align: center;
                color: #333;
            }}
        </style>
    </head>
    <body>
        <div class="dashboard">
            <h1>{title}</h1>
            <div class="image-grid">
    """
    
    # Agregar imágenes al HTML
    for img_path in image_paths:
        html_content += f"""
                <div class="image-container">
                    <img src="{img_path}" alt="Dashboard Image">
                </div>
        """
    
    html_content += """
            </div>
        </div>
    </body>
    </html>
    """
    
    # Guardar el archivo HTML
    output_file = "dashboard.html"
    with open(output_file, "w") as f:
        f.write(html_content)
    
    return output_file
```

```python
imagenes = [
    "ruta/imagen1.png",
    "ruta/imagen2.png",
    "ruta/imagen3.png",
    "ruta/imagen4.png"
]

# Crear el dashboard
dashboard_file = create_image_dashboard(
    title="Mi Dashboard de Imágenes",
    image_paths=imagenes,
    columns=2  # Mostrar 2 imágenes por fila
)

print(f"Dashboard creado: {dashboard_file}")
```
